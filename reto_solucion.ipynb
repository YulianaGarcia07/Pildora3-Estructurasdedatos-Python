## Datos de partida
Ejecuta esta celda primero. **No la modifiques.**

```python
# Lista de tuplas: (título, artista, duración en segundos)
canciones = [
    ("Blinding Lights", "The Weeknd", 200),
    ("Shape of You", "Ed Sheeran", 234),
    ("Starboy", "The Weeknd", 230),
    ("Perfect", "Ed Sheeran", 263),
    ("Bad Guy", "Billie Eilish", 194),
]

# Lista con los títulos de una playlist
playlist = ["Blinding Lights", "Shape of You", "Starboy", "Perfect", "Bad Guy"]

# Lista con el artista de cada canción (hay repetidos)
artistas_lista = ["The Weeknd", "Ed Sheeran", "The Weeknd", "Ed Sheeran", "Billie Eilish"]

# Diccionario: título -> número de reproducciones
reproducciones = {
    "Blinding Lights": 120,
    "Shape of You": 95,
    "Starboy": 80,
    "Perfect": 60,
    "Bad Guy": 150,
}

# Diccionario de listas: usuario -> sus canciones
playlists = {
    "Ana": ["Blinding Lights", "Starboy", "Bad Guy", "Perfect"],
    "Luis": ["Shape of You", "Perfect", "Bad Guy"],
}

# Cola de reproducción con canciones repetidas
cola = ["Starboy", "Bad Guy", "Starboy", "Perfect", "Bad Guy", "Starboy"]
```

---
# 🟢 Nivel 1 · Calentamiento

## 1.1 · Lista
1. Añade `"Levitating"` al final de `playlist`.
2. Ordena `playlist` alfabéticamente.
3. Imprime la **primera** canción, la **última** y **cuántas** hay.

> 💡 Pista: `append`, `sort`, índices `[0]` y `[-1]`, y `len`.

```python
# append añade UN elemento al final de la lista
playlist.append("Levitating")

# sort ordena alfabéticamente (modifica la propia lista)
playlist.sort()

# [0] es la primera, [-1] es la última, len() cuenta los elementos
print(playlist[0], playlist[-1], len(playlist))
```

**Resultado esperado:** `Bad Guy Starboy 6`

## 1.2 · Tupla
1. **Desempaqueta** `canciones[0]` en tres variables: `titulo`, `artista` y `seg`.
2. Imprime el título, el artista y la duración **en minutos**, redondeada a 1 decimal.

> 💡 Pista: `a, b, c = tupla` desempaqueta. Para pasar de segundos a minutos, divide entre 60 y usa `round(valor, 1)`.

```python
# Desempaquetar: cada variable recibe un elemento de la tupla, en orden
titulo, artista, seg = canciones[0]

# Dividimos entre 60 para pasar a minutos y redondeamos a 1 decimal
print(titulo, "-", artista, "-", round(seg / 60, 1), "min")
```

**Resultado esperado:** `Blinding Lights - The Weeknd - 3.3 min`

## 1.3 · Set
1. Convierte `artistas_lista` en un set llamado `artistas`.
2. Imprime cuántos artistas **distintos** hay.
3. Comprueba si `"Drake"` está en `artistas`.

> 💡 Pista: `set()`, `len` e `in`.

```python
# set() elimina los repetidos automáticamente
artistas = set(artistas_lista)

print(len(artistas))          # cuántos artistas distintos
print("Drake" in artistas)    # in devuelve True o False
```

**Resultado esperado:** `3` y `False`

# 🟡 Nivel 2 · Reto principal

## 2.1 · Diccionario
1. Suma **10 reproducciones** a `"Bad Guy"`.
2. Añade la canción `"Levitating"` con **40** reproducciones.
3. Imprime el **total** de reproducciones.

> 💡 Pista: `dic[clave] += 10`, `dic[clave_nueva] = valor` y `sum(dic.values())`.

```python
# La clave ya existe: se modifica su valor
reproducciones["Bad Guy"] += 10

# La clave no existe: se añade
reproducciones["Levitating"] = 40

# values() devuelve los valores y sum() los suma
print(sum(reproducciones.values()))
```

**Resultado esperado:** `555`

## 2.2 · Diccionario de listas
El diccionario `playlists` guarda la lista de canciones de cada usuario.
1. **Luis** añade `"Levitating"` a su lista.
2. Imprime la **segunda** canción de **Ana**.
3. Imprime **cuántas** canciones tiene Luis.

> 💡 Pista: primero la clave y luego lo que quieras hacer con la lista: `playlists["Luis"].append(...)`, `playlists["Ana"][...]`.

```python
# playlists["Luis"] es una lista, así que podemos usar append sobre ella
playlists["Luis"].append("Levitating")

# Primero la clave, luego el índice (la segunda canción es la posición 1)
print(playlists["Ana"][1])

# len sobre la lista de Luis
print(len(playlists["Luis"]))
```

**Resultado esperado:** `Starboy` y `4`

## 2.3 · Sets entre usuarios
Convierte la lista de cada usuario en un set: `ana` y `luis`. Después imprime:
1. Las canciones que tienen **en común**.
2. Las canciones que **solo** tiene Ana.
3. **Cuántas** canciones distintas hay entre los dos.

> 💡 Pista: `&` (en común), `-` (diferencia) y `|` (todas, sin repetir).

```python
# Convertimos cada lista en un set para poder compararlas
ana = set(playlists["Ana"])
luis = set(playlists["Luis"])

print(ana & luis)          # en común
print(ana - luis)          # solo en Ana
print(len(ana | luis))     # todas las canciones distintas
```

**Resultado esperado:** `{'Bad Guy', 'Perfect'}`, `{'Blinding Lights', 'Starboy'}` y `6` (el orden dentro de un set puede variar)

## 2.4 · Consultar con seguridad
1. Obtén las reproducciones de `"Drake"` **sin que dé error**: si no existe, que devuelva `0`.
2. Imprime el **mayor** número de reproducciones de todas las canciones.

> 💡 Pista: `get` con un valor por defecto, y `max(dic.values())`.

```python
# get con un segundo argumento evita el KeyError: devuelve 0 si la clave no existe
print(reproducciones.get("Drake", 0))

# max sobre los valores devuelve el mayor
print(max(reproducciones.values()))
```

**Resultado esperado:** `0` y `160`

# 🔴 Nivel 3 · Desafío

## 3.1 · Estructuras anidadas
Desempaqueta `canciones[1]` en `titulo`, `artista` y `seg`, y crea un diccionario `ficha` con estas claves:
- `"titulo"` y `"artista"`.
- `"duracion"`: una **tupla** `(minutos, segundos)`.
- `"reproducciones"`: sacadas del diccionario `reproducciones`.

Imprime `ficha` y después solo los **segundos** de la duración.

> 💡 Pista: `seg // 60` da los minutos completos y `seg % 60` los segundos que sobran.

```python
titulo, artista, seg = canciones[1]

# Un diccionario cuyo valor "duracion" es una tupla
ficha = {
    "titulo": titulo,
    "artista": artista,
    "duracion": (seg // 60, seg % 60),      # // = minutos completos, % = segundos sobrantes
    "reproducciones": reproducciones[titulo],
}

print(ficha)
print(ficha["duracion"][1])                 # clave y luego índice
```

**Resultado esperado:** `{'titulo': 'Shape of You', 'artista': 'Ed Sheeran', 'duracion': (3, 54), 'reproducciones': 95}` y `54`

## 3.2 · Limpiar duplicados
La lista `cola` tiene canciones repetidas.
1. Crea `sin_repetidos`: una **lista** con cada canción una sola vez.
2. Imprime cuántas canciones distintas hay y **cuántas repeticiones** se eliminaron.

> 💡 Pista: `list(set(...))` y `len`.

```python
# Pasamos por un set para quitar repetidos y volvemos a lista
sin_repetidos = list(set(cola))

# Eliminadas = total original - total sin repetidos
print(len(sin_repetidos), len(cola) - len(sin_repetidos))
```

**Resultado esperado:** `3 3`

## 3.3 · Resumen final
Crea una **tupla** `resumen` con, en este orden:
1. Cuántas canciones hay en `playlist`.
2. Cuántos artistas distintos hay en `artistas`.
3. El total de reproducciones.

Después **desempaquétala** en `n_canciones`, `n_artistas` y `n_reprod`, e imprime `resumen`.

```python
# Una tupla agrupa los tres resultados en un dato que no debería cambiar
resumen = (len(playlist), len(artistas), sum(reproducciones.values()))

n_canciones, n_artistas, n_reprod = resumen
print(resumen)
```

**Resultado esperado:** `(6, 3, 555)`

## 3.4 · ¿Qué estructura usarías?
| Caso | Estructura | Por qué |
|---|---|---|
| a) Fecha de lanzamiento (día, mes, año) | **Tupla** | Son datos fijos que no deben cambiar |
| b) Reproducciones de cada canción | **Diccionario** | Se busca por el nombre de la canción (clave) |
| c) Géneros sin repetir | **Set** | Elimina duplicados automáticamente |
| d) Cola de reproducción | **Lista** | Es ordenada y se pueden añadir elementos |
