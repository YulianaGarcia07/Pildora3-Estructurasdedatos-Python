## ¿Qué es una estructura de datos?

Una **estructura de datos** es un conjunto de datos guardado bajo un mismo nombre.
En Python hay cuatro básicas, y cada una se parece a algo cotidiano:

| Estructura | Analogía | Idea clave |
|---|---|---|
| **Lista** | Lista de la compra | Ordenada y modificable |
| **Tupla** | Coordenadas de un punto | Fija, no se puede modificar |
| **Set** | Conjunto de elementos distintos | Sin repetidos y sin orden |
| **Diccionario** | Agenda del móvil | Buscas por nombre (clave) y obtienes el dato (valor) |

Para cada una responderemos las mismas **8 preguntas**: sintaxis, orden e indexación,
duplicados y anidados, mutable, dinámica, iterable, métodos y casos de uso.

# 1. Listas `[ ]`

| Pregunta | Respuesta |
|---|---|
| Sintaxis | `[ ]` o `list()` |
| Orden e indexación | Ordenada, índices desde 0 |
| Duplicados y anidados | Sí / Sí |
| Mutable | Sí |
| Dinámica (añadir o quitar) | Sí |
| Iterable | Sí, con `for` |
| Métodos principales | `append`, `extend`, `insert`, `remove`, `pop`, `index`, `count`, `sort`, `reverse` |
| Casos de uso | Pedidos en orden de llegada, resultados que se van acumulando |

### 1.1 Crear una lista y acceder a sus elementos

```python
# Creamos una lista con corchetes; los elementos van separados por comas
pedidos = ["P001", "P002", "P003"]

# type() nos dice de qué tipo es la variable
print(type(pedidos))

# Los índices empiezan en 0: el primero es [0] y el último es [-1]
print(pedidos[0], pedidos[-1])
```

### 1.2 Modificar un elemento (las listas son mutables)

```python
# Asignamos un valor nuevo a la posición 1
pedidos[1] = "P020"
print(pedidos)
```

### 1.3 `append`: añade UN elemento al final

```python
pedidos.append("P004")
print(pedidos)

# Ojo: si le pasamos otra lista, la mete COMO UN SOLO elemento (lista anidada)
pedidos.append(["P005", "P006"])
print(pedidos)

# Para llegar a "P005" usamos dos pares de corchetes: [posición de fuera][posición de dentro]
print(pedidos[-1][0])
```

### 1.4 `extend`: añade CADA elemento por separado

```python
# Volvemos a una lista limpia para que se vea bien la diferencia con append
pedidos = ["P001", "P002", "P003", "P004"]

pedidos.extend(["P005", "P006"])
print(pedidos)
```

### 1.5 `insert`: añade en una posición concreta

```python
# insert(posición, valor): lo demás se corre un lugar a la derecha
pedidos.insert(1, "P010")
print(pedidos)
```

### 1.6 Eliminar elementos: `remove`, `del` y `pop`

```python
pedidos.remove("P003")       # borra por VALOR
del pedidos[0]               # borra por POSICIÓN
ultimo = pedidos.pop()       # quita el último y nos lo devuelve

print(pedidos)
print("El último que quitamos fue:", ultimo)
```

### 1.7 Consultar: `index`, `count`, `in` y `len`

```python
print(pedidos.index("P004"))     # en qué posición está
print(pedidos.count("P004"))     # cuántas veces aparece
print("P004" in pedidos)         # ¿está en la lista? (True o False)
print(len(pedidos))              # cuántos elementos tiene
```

### 1.8 Ordenar y cortar trozos (slicing)

```python
pedidos.sort()                   # ordena de menor a mayor
print(pedidos)

pedidos.sort(reverse=True)       # ordena al revés
print(pedidos)

pedidos.reverse()                # solo invierte el orden actual
print(pedidos)

# Slicing [inicio:fin]: el fin NO se incluye
print(pedidos[1:3])              # del índice 1 al 2
print(pedidos[-2:])              # los dos últimos
```

### 1.9 Iterables: recorrer una lista (adelanto del `for`)

*Iterable* significa que se puede recorrer elemento por elemento.
El `for` toma los elementos de uno en uno: en cada vuelta, `p` vale un elemento distinto.
Los bucles se verán en detalle más adelante.

```python
for p in pedidos:
    print(p)
```

**Resumen de listas:** ordenadas, modificables, admiten repetidos y anidados.
`append` añade uno, `extend` añade varios, `remove` borra por valor y `del` / `pop` por posición.

# 2. Tuplas `( )`

| Pregunta | Respuesta |
|---|---|
| Sintaxis | `( )` o `tuple()`; con un solo elemento: `(1,)` |
| Orden e indexación | Ordenada, índices desde 0 |
| Duplicados y anidados | Sí / Sí |
| Mutable | **No** |
| Dinámica (añadir o quitar) | **No**, tamaño fijo |
| Iterable | Sí |
| Métodos principales | Solo `index` y `count` |
| Casos de uso | Coordenadas, fechas, datos que no deben cambiar |

### 2.1 Crear una tupla y acceder

```python
# Una tupla se crea con paréntesis
ubicacion = (40.4168, -3.7038)

print(ubicacion[0])                  # se accede igual que en una lista
print(len(ubicacion))                # tamaño
print(ubicacion.index(-3.7038))      # posición de un valor
print(ubicacion.count(40.4168))      # cuántas veces aparece
```

### 2.2 ⚠️ Intentar modificarla da error (esa es su ventaja)

```python
ubicacion[0] = 41.0
```

### 2.3 Detalles: un solo elemento, convertir una lista y ocupar menos memoria

```python
# Con un solo elemento hace falta una coma; sin ella no es una tupla
unico = ("a",)
no_es_tupla = ("a")
print(type(unico), type(no_es_tupla))

# Convertir una lista en tupla
pedidos_lista = ["P001", "P002", "P003"]
pedidos_tupla = tuple(pedidos_lista)
print(pedidos_tupla)

# Una tupla ocupa menos memoria que una lista con los mismos datos
import sys
print("Lista: ", sys.getsizeof([1, 2, 3]), "bytes")
print("Tupla: ", sys.getsizeof((1, 2, 3)), "bytes")
```

### 2.4 Iterables

```python
for coordenada in ubicacion:
    print(coordenada)
```

**Resumen de tuplas:** como una lista, pero inmutable. Protege los datos y ocupa menos memoria.
Solo tiene dos métodos: `index` y `count`.

# 3. Sets `{ }`

| Pregunta | Respuesta |
|---|---|
| Sintaxis | `{ }` o `set()`; **el set vacío es `set()`** (`{}` crea un diccionario) |
| Orden e indexación | **Sin orden, sin índices** |
| Duplicados y anidados | **No** admite duplicados / Solo elementos inmutables (no listas) |
| Mutable | Sí |
| Dinámica (añadir o quitar) | Sí |
| Iterable | Sí (el orden no está garantizado) |
| Métodos principales | `add`, `remove`, `discard`, `pop`, `union`, `intersection`, `difference` |
| Casos de uso | Eliminar duplicados, comprobar pertenencia, comparar grupos |

### 3.1 Crear un set a partir de una lista (se eliminan los repetidos)

```python
ciudades_lista = ["Madrid", "Sevilla", "Madrid", "Valencia", "Sevilla"]

# set() convierte la lista en un conjunto: los duplicados desaparecen
ciudades = set(ciudades_lista)

print(ciudades)          # ojo: el orden puede cambiar
print(len(ciudades))
```

### 3.2 Añadir elementos con `add`

```python
ciudades.add("Bilbao")
ciudades.add("Madrid")    # ya existe: no se duplica ni da error
print(ciudades)
```

### 3.3 ⚠️ Un set no tiene índices

```python
ciudades[0]
```

### 3.4 Quitar elementos: `discard` frente a `remove`

```python
# discard NO da error si el elemento no existe
ciudades.discard("Murcia")
print(ciudades)
```

```python
# remove SÍ da error si el elemento no existe
ciudades.remove("Murcia")
```

### 3.5 Caso práctico: quitar duplicados de una lista

```python
sin_duplicados = list(set(ciudades_lista))
print(sin_duplicados)
# Ojo: al pasar por un set se pierde el orden original
```

### 3.6 Iterables

```python
for ciudad in ciudades:
    print(ciudad)
```

**Resumen de sets:** elementos únicos, sin orden y sin índices.
`discard` no da error si no existe y `remove` sí. Su uso estrella es eliminar duplicados.

# 4. Diccionarios `{clave: valor}`

| Pregunta | Respuesta |
|---|---|
| Sintaxis | `{clave: valor}` o `dict()`; el vacío es `{}` |
| Orden e indexación | Conserva el orden de inserción; se accede por **clave**, no por posición |
| Duplicados y anidados | Claves **no**, valores sí / Sí, los valores pueden ser de cualquier tipo |
| Mutable | Sí |
| Dinámica (añadir o quitar) | Sí |
| Iterable | Sí (por defecto recorre las claves) |
| Métodos principales | `get`, `keys`, `values`, `items`, `update`, `pop` |
| Casos de uso | Buscar por ID, contar, configuraciones, datos tipo JSON |

### 4.1 Crear un diccionario y acceder por clave

```python
# Cada elemento es un par clave: valor
stock = {"portatil": 10, "raton": 20}

print(stock["portatil"])     # accedemos con la clave, no con una posición
print(len(stock))            # cuántos pares tiene
```

### 4.2 ⚠️ Una clave que no existe da `KeyError`

```python
stock["monitor"]
```

### 4.3 `get` e `in`

```python
print(stock.get("monitor"))        # None, sin error
print(stock.get("monitor", 0))     # 0, el valor por defecto que elegimos
print("raton" in stock)            # in busca en las CLAVES
print(20 in stock)                 # los valores no cuentan
```

### 4.4 Añadir y modificar

```python
stock["teclado"] = 8       # clave nueva: se añade
stock["raton"] -= 5        # clave existente: se modifica
print(stock)
```

### 4.5 `update` y `pop`

```python
# update mete todos los pares de otro diccionario
stock.update({"monitor": 4})
print(stock)

# pop saca una clave y nos devuelve su valor
vendido = stock.pop("monitor")
print(vendido)
print(stock)
```

### 4.6 Claves, valores, pares e iterables

```python
print(stock.keys())      # las claves
print(stock.values())    # los valores
print(stock.items())     # los pares: cada par es una tupla
```

```python
# Por defecto, el for recorre las CLAVES
for clave in stock:
    print(clave)

# Con items() recibimos la clave y el valor a la vez
for clave, valor in stock.items():
    print(clave, valor)
```

### 4.7 Todo junto: estructuras dentro de estructuras

```python
cliente = {
    "nombre": "Ana",
    "pedidos": ["P001", "P002"],        # el valor es una lista
    "ubicacion": (40.41, -3.70),        # el valor es una tupla
}

# Primero la clave y luego el índice
print(cliente["pedidos"][0])
print(cliente["ubicacion"][1])
```

**Resumen de diccionarios:** pares clave-valor, claves únicas, se accede por clave.
`get` evita el `KeyError`. Es la estructura más usada en análisis de datos.

# 5. Comparativa y regla de decisión

| | Lista | Tupla | Set | Diccionario |
|---|---|---|---|---|
| Sintaxis | `[ ]` | `( )` | `{ }` | `{k: v}` |
| Orden | Sí | Sí | No | Sí (inserción) |
| Duplicados | Sí | Sí | No | Claves no |
| Mutable | Sí | No | Sí | Sí |

## ¿Qué estructura uso?
1. ¿Necesito **buscar por un nombre o un ID**? → **Diccionario**
2. ¿Necesito que **no haya repetidos**? → **Set**
3. ¿Los datos **no deben cambiar nunca**? → **Tupla**
4. Si no es ninguno de los casos anteriores → **Lista**

**Ejemplos:** stock por producto (diccionario) · ciudades de envío sin repetir (set) ·
latitud y longitud del almacén (tupla) · pedidos del día en orden de llegada (lista).

# 6. Coste computacional: buscar en una lista frente a un set

En una lista, `in` mira los elementos **de uno en uno**. En un set o un diccionario, va casi directo.
Con pocos datos no se nota; con millones, sí.

```python
import time

grande_lista = list(range(5_000_000))
grande_set = set(grande_lista)

# Buscamos el último número: el peor caso para la lista
inicio = time.perf_counter()
4_999_999 in grande_lista
tiempo_lista = time.perf_counter() - inicio

inicio = time.perf_counter()
4_999_999 in grande_set
tiempo_set = time.perf_counter() - inicio

# Pasamos los segundos a milisegundos (x 1000) y redondeamos a 3 decimales
print("Lista:", round(tiempo_lista * 1000, 3), "milisegundos")
print("Set:  ", round(tiempo_set * 1000, 3), "milisegundos")
```

# 7. Primeros auxilios: errores comunes

| Error | Qué significa | Cómo solucionarlo |
|---|---|---|
| `TypeError: 'tuple' object does not support item assignment` | Las tuplas no se modifican | Conviértela a lista con `list()`, o crea otra tupla |
| `TypeError: 'set' object is not subscriptable` | Los sets no tienen índices | Conviértelo con `list(mi_set)` |
| `KeyError: 'clave'` | La clave no existe en el diccionario | Usa `dic.get("clave", valor_por_defecto)` |
| `IndexError: list index out of range` | El índice es mayor que el tamaño de la lista | Comprueba el tamaño con `len()` |
| `ValueError: list.remove(x): x not in list` | Intentas quitar un valor que no está | Comprueba antes con `in` |
| `TypeError: unhashable type: 'list'` | Una lista no puede ser clave ni elemento de un set | Usa una tupla |
| `{}` no crea un set | Crea un diccionario vacío | Usa `set()` |
| `("a")` no es una tupla | Son solo paréntesis | Añade la coma: `("a",)` |
| Cambias `b` y también cambia `a` | `b = a` no copia, apunta a la misma lista | Usa `b = a.copy()` |
